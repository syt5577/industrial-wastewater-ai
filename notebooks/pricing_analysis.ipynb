```python
import pandas as pd
from pathlib import Path
import os

file_path = Path.cwd().parent/'data'/'construct'/'price_history.xlsx'
print(file_path)
```

```python
df = pd.read_excel(file_path)
print(df.dtypes)
print(df.head())
```

```python
print(df.columns.tolist())
```

```python
base_cost = 8.93

limits = {
    "COD_mg_L": 600,
    "CODS_mg_L": 300,
    "SS_mg_L": 200,
    "NH4_N_mg_L": 30,
    "TN_mg_L": 70,
    "TP_mg_L": 3.2,
    "TDS_mg_L": 1600,
    "BOD5_mg_L": 250,
    "TOC_mg_L": 150,
    "OIL_mg_L": 30,
}
```

```python
for column,limit in limits.items():
    ratio_column = column.replace('_mg_L','_ratio')
    df[ratio_column] = df[column] / limit
    print (df[ratio_column])
```

```python
ratio_columns = [column.replace('_mg_L','_ratio') for column in limits ]
print(ratio_columns)
```

```python
print(df[['case_id','customer_name']+ratio_columns])
```

```python
ratio_df = df[ratio_columns]
print(ratio_df)
```

```python
df['max_ratio'] = ratio_df.max(axis=1)
df['max_ratio_indicator'] = ratio_df.idxmax(axis=1).str.replace('_ratio','',regex = False)
df['is_exceeded'] = df['max_ratio'] >1
```

```python
df[[
    "case_id",
    "customer_name",
    "max_ratio",
    "max_ratio_indicator",
    "is_exceeded"
]]
```

```python
df['ph_exception'] = (df['pH']<6) | (df['pH']>9)
print(df['ph_exception'])
```

```python
base_cost = 8.93
df['rule_price'] = df['max_ratio'].apply(lambda x :base_cost*max(1.2,x))
print(df['rule_price'])
```

```python
print(df[['max_ratio', 'rule_price']])
```

```python
price_columns = [
    'case_id',
    'customer_name',
    'COD_mg_L',
    'NH4_N_mg_L',
    'TP_mg_L',
    'initial_quote_yuan_per_t',
    'agreed_price_yuan_per_t',
    'actual_settlement_price_yuan_per_t'
]

history_df = df[price_columns].copy()
```

```python
print(history_df.to_string(index=False))
```

```python
new_sample = {
    'COD_mg_L': 8000,
    'NH4_N_mg_L': 500,
    'TP_mg_L': 30
}
```

```python
history_df['COD_diff'] = (history_df['COD_mg_L']-new_sample['COD_mg_L']).abs()/max(new_sample['COD_mg_L'],1e-6)
print(history_df[['customer_name', 'COD_diff']])
```

```python
features = ['COD_mg_L', 'NH4_N_mg_L', 'TP_mg_L']

for column in features:
    diff_column = column.replace('_mg_L', '_diff')

    history_df[diff_column] = (
        (history_df[column] - new_sample[column]).abs()
        / max(new_sample[column], 1e-6)
    )
```

```python
print(history_df[['customer_name', 'COD_diff', 'NH4_N_diff', 'TP_diff']])
```

```python
diff_columns = ['COD_diff', 'NH4_N_diff', 'TP_diff']

history_df['distance'] = history_df[diff_columns].mean(axis=1)
```

```python
top3 = history_df.sort_values('distance').head(3)

print(top3[
    ['case_id', 'customer_name', 'distance',
     'initial_quote_yuan_per_t',
     'agreed_price_yuan_per_t',
     'actual_settlement_price_yuan_per_t']
].to_string(index=False))
```

```python
best_case = top3.iloc[0]

reference_price = best_case['actual_settlement_price_yuan_per_t']

print('最相似客户：', best_case['customer_name'])
print('历史参考价格：', reference_price)
```

```python
from pydantic import BaseModel,Field
from typing import Optional

class WaterQuality(BaseModel):
    COD_mg_L :Optional[float]=Field(
        default = None,
        description = 'COD,cod,化学需氧量，单位mg/L'
    )
    NH4_N_mg_L: Optional[float] = Field(
        default=None,
        description="氨氮、NH4-N，单位 mg/L"
    )

    TP_mg_L: Optional[float] = Field(
        default=None,
        description="总磷、TP，单位 mg/L"
    )
    
```

```python
print(WaterQuality.model_json_schema())
```

```python
from dotenv import load_dotenv
import os
from langchain_openai import ChatOpenAI
from pathlib import Path
env_path = Path.cwd().parent / ".env"
load_dotenv(env_path)
api_key = os.getenv("SILICONFLOW_API_KEY")

llm = ChatOpenAI(
    api_key=api_key,
    base_url="https://api.siliconflow.cn/v1",
    model="deepseek-ai/DeepSeek-V3.2"
)
```

```python
structured_llm = llm.with_structured_output(WaterQuality)
```

```python
response = structured_llm.invoke("帮我算一下污水价格，氨氮300，cod500，总磷30")
print(response)
```

```python
water_quality = response.model_dump(exclude_none=True)
print(water_quality)
print(type(water_quality))
```

```python
limits = {
    "COD_mg_L": 600,
    "CODS_mg_L": 300,
    "SS_mg_L": 200,
    "NH4_N_mg_L": 30,
    "TN_mg_L": 70,
    "TP_mg_L": 3.2,
    "TDS_mg_L": 1600,
    "BOD5_mg_L": 250,
    "TOC_mg_L": 150,
    "OIL_mg_L": 30,
}
base_cost = 8.93
```

```python
ratio = {}
for indicator,value in water_quality.items():
    if indicator in limits:
        ratio[indicator] = value/limits[indicator]
print(ratio)
```

```python
max_indicator = max(ratio,key = ratio.get)
max_ratio = ratio[max_indicator]
rule_price = base_cost * max(1.2,max_ratio)
```

```python
print("最大超标指标：", max_indicator)
print("最大超标倍数：", max_ratio)
print("制度规则价格：", round(rule_price, 2), "元/吨")
```
