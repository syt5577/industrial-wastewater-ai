```python
from pydantic import BaseModel,Field
from typing import Optional

class WaterQuality(BaseModel):
    COD_mg_L :Optional[float]=Field(
        default = None,
        description = 'COD,cod,化学需氧量，单位mg/L'
    )
    NH4_N_mg_L: Optional[float] = Field(
        default=None,
        description="氨氮、NH4-N，单位 mg/L"
    )

    TP_mg_L: Optional[float] = Field(
        default=None,
        description="总磷、TP，单位 mg/L"
    )
    
```

```python
print(WaterQuality.model_json_schema())
```

```python
def analyze_wastewater_price(water_quality, history_df, limits, base_cost):
    if not water_quality:
        raise ValueError('未提供水质报价、无法计算报价')
    if any(value<0 for value in water_quality.values()):
        raise ValueError('水质指标不能为负')
    pricing_indicators = [indicator for indicator in water_quality if indicator in limits]
    unpriced_indicators = [indicator for indicator in water_quality if indicator not in limits]
    ratios = {}
    for indicator in pricing_indicators:
        value = water_quality[indicator]
        ratios[indicator] = value/limits[indicator]
    if ratios:
        max_indicator = max(ratios,key = ratios.get)
        max_ratio = ratios[max_indicator]
        rule_price = round(base_cost * max(1.2,max_ratio),2)
    if not ratios:
        max_indicator = None
        max_ratio = None
        rule_price = None
    features = [
        column for column in water_quality
        if column in history_df.columns and column in limits
    ]
    if features:
        diff_df = pd.DataFrame(index=history_df.index)
        for column in features:
            diff_column = column.replace('_mg_L','_diff')
            diff_df[diff_column] = (history_df[column]-water_quality[column]).abs()/max(water_quality[column],1e-6)    
        diff_df['distance'] = diff_df.mean(axis=1)
        top3 = (diff_df.sort_values('distance')
            .head(3))
        price_columns = [
        'case_id',
        'customer_name',
        'initial_quote_yuan_per_t',
        'agreed_price_yuan_per_t',
        'actual_settlement_price_yuan_per_t'
            ]
        historical_cases = history_df.loc[top3.index,price_columns].copy()
        historical_cases['distance'] = top3['distance']
        historical_cases = historical_cases.astype(object).where(
        pd.notna(historical_cases),None
        )
        historical_cases = historical_cases.to_dict(orient = 'records')

    else:
        historical_cases = []
    
    return {
    'input_water_quality': water_quality,
    'ratios': ratios,
    'max_indicator': max_indicator,
    'max_ratio': max_ratio,
    'rule_price': rule_price,
    'historical_cases': historical_cases,
    'unpriced_indicators': unpriced_indicators
}
```

```python
pricing_result = analyze_wastewater_price(
    water_quality,
    history_df,
    limits,
    base_cost
)

print(pricing_result)
```
