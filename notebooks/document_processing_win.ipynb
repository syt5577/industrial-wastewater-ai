```python
def hybrid_search(query,k=5,rrf_k = 60):
    bm25_results =bm25_retriever.invoke(query)
    vectore_results = vectorstore.similarity_search(query,k=k)
    scores = {}
    docs = {}
    for rank,doc in enumerate(bm25_results,start=1):
        chunk_id = doc.metadata["chunk_id"]
        socre = scores.get(chunk_id,0) + 1/(rank+rrf_k)
        scores[chunk_id] = socre
        docs[chunk_id] = doc
    for rank,doc in enumerate(vectore_results,start=1):
        chunk_id = doc.metadata["chunk_id"]
        socre = scores.get(chunk_id,0) + 1/(rank+rrf_k)
        scores[chunk_id] = socre
        docs[chunk_id] = doc
    rank_doc = sorted(scores,key = scores.get ,reverse= True)
    return [docs[chunk_id] for chunk_id in rank_doc[:k]]
```

```python
def hybrid_search(query, k=10):
    vector_docs = vector_retriever.invoke(query)
    bm25_docs = bm25_retriever.invoke(query)

    scores = {}
    doc_map = {}

    for docs in [vector_docs, bm25_docs]:
        for rank, doc in enumerate(docs, start=1):
            chunk_id = doc.metadata["chunk_id"]

            if chunk_id not in scores:
                scores[chunk_id] = 0
                doc_map[chunk_id] = doc

            scores[chunk_id] += 1 / (60 + rank)

    ranked_ids = sorted(
        scores,
        key=lambda chunk_id: scores[chunk_id],
        reverse=True
    )

    return [
        (doc_map[chunk_id], scores[chunk_id])
        for chunk_id in ranked_ids[:k]
    ]
questions = [
    "外委污水处理业务由哪些部门负责？",
    "外委污水处理业务职责与分工",
    "发展与市场部、生产部、安全环保部在外委污水业务中的职责"
]

for question in questions:
    print("\n问题：", question)

    results = hybrid_search(question, k=10)

    for rank, (doc, score) in enumerate(results, start=1):
        print(
            rank,
            doc.metadata["chunk_id"],
            doc.metadata["subsection"],
            round(score, 4)
        )
```
